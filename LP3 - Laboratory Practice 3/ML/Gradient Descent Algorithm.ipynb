Implement Gradient Descent Algorithm to find the local minima of a function. <br>
For example, find the local minima of the function <br>
y=(x+3)² starting from the point x=2. <br>

```python

import numpy as np
import matplotlib.pyplot as plt

# Define the function
def f(x):
    return (x + 3) ** 2

# Define the derivative
def df(x):
    return 2 * (x + 3)

# Initialize parameters
x = 2.0
learning_rate = 0.1
iterations = 50

# Store values for visualization
x_history = [x]
y_history = [f(x)]

print("Iteration\t x\t\t Gradient\t y = (x+3)^2")
print("-" * 65)

# Gradient Descent algorithm
for i in range(iterations):
    gradient = df(x)
    y = f(x)

    print(f"{i:3d}\t\t{x:.6f}\t{gradient:.6f}\t{y:.6f}")

    # Update x
    x = x - learning_rate * gradient

    x_history.append(x)
    y_history.append(f(x))

# Print final result
print("-" * 65)
print(f"Final x value: {x:.6f}")
print(f"Final y value: {f(x):.10f}")
```

```python
# Plot the function and descent path
x_values = np.linspace(-5, 3, 200)

plt.figure(figsize=(8, 5))
plt.plot(x_values, f(x_values), label="y = (x + 3)^2")
plt.scatter(x_history, y_history, color="red",
            s=18, label="Gradient Descent steps")
plt.scatter([-3], [0], color="green",
            s=70, label="Minimum (-3, 0)")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Gradient Descent Algorithm")
plt.legend()
plt.grid(True)
plt.show()
```
